#### SQL Query

```python
sql_query = \
"""
SELECT
    FORMAT_DATE('%Y-%m-%d', purchase_date) AS ds,
    ROUND(SUM(final_price), 2) AS y
FROM
 `gcp-mts-pf.ls_customers.tb_sales`
GROUP BY
    ALL
ORDER BY 1;
"""
```

```python
def read_bq(sql_query: str, ) -> pl.DataFrame:

    if not isinstance(sql_query, str):
        raise ValueError("The 'query' parameter must be a non-empty string.")

    _client = bigquery.Client(project="gcp-mts-pf")

    try:
        logging.info("Executing BigQuery query...")
        job_config = QueryJobConfig()
        job_config.use_legacy_sql = False

        query_job = _client.query(sql_query, job_config=job_config)
        rows = query_job.to_arrow()
        logging.info("Query executed successfully.")

    except Exception as e:
        logging.error(f"Error executing query: {e}")
        raise

    return pl.from_arrow(rows)
```

```python
df_result = read_bq(sql_query=sql_query)
```

Por que yearly_seasonality=False?

Porque temos apenas aproximadamente 6 meses de histórico. Não temos dados suficientes para aprender uma sazonalidade anual de forma confiável.

E daily_seasonality=False porque nossa granularidade é diária. Não temos várias observações dentro do mesmo dia para aprender um padrão intradiário.

```python
import pandas as pd
import numpy as np
import polars as pl

from prophet import Prophet
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)
```

```python
horizon = 28

cutoffs = [
    pd.Timestamp("2026-06-06"),
    pd.Timestamp("2026-06-20"),
    pd.Timestamp("2026-07-07"),
]
```

```python
def evaluate_prophet(
    train: pd.DataFrame,
    test: pd.DataFrame
) -> dict:

    model = Prophet(
        weekly_seasonality=True,
        daily_seasonality=False,
        yearly_seasonality=False
    )

    model.fit(train)

    future = model.make_future_dataframe(
        periods=len(test),
        freq="D"
    )

    forecast = model.predict(future)

    forecast_test = (
        forecast[
            ["ds", "yhat"]
        ]
        .tail(len(test))
    )

    comparison = test.merge(
        forecast_test,
        on="ds",
        how="left"
    )

    mae = mean_absolute_error(
        comparison["y"],
        comparison["yhat"]
    )

    rmse = np.sqrt(
        mean_squared_error(
            comparison["y"],
            comparison["yhat"]
        )
    )

    mape = mean_absolute_percentage_error(
        comparison["y"],
        comparison["yhat"]
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "MAPE": mape
    }
```

```python
results = []

for cutoff in cutoffs:

    train = df_prophet[
        df_prophet["ds"] <= cutoff
    ].copy()

    test_end = cutoff + pd.Timedelta(days=horizon)

    test = df_prophet[
        (df_prophet["ds"] > cutoff) &
        (df_prophet["ds"] <= test_end)
    ].copy()

    metrics = evaluate_prophet(
        train=train,
        test=test
    )

    results.append({
        "cutoff": cutoff,
        "train_rows": len(train),
        "test_rows": len(test),
        **metrics
    })
```

```python
def evaluate_baseline(
    train: pd.DataFrame,
    test: pd.DataFrame
) -> dict:

    prediction = train["y"].mean()

    y_pred = np.full(
        len(test),
        prediction
    )

    mae = mean_absolute_error(
        test["y"],
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            test["y"],
            y_pred
        )
    )

    mape = mean_absolute_percentage_error(
        test["y"],
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "MAPE": mape
    }
```

```python
baseline_results = []

for cutoff in cutoffs:

    train = df_prophet[
        df_prophet["ds"] <= cutoff
    ].copy()

    test_end = cutoff + pd.Timedelta(days=horizon)

    test = df_prophet[
        (df_prophet["ds"] > cutoff) &
        (df_prophet["ds"] <= test_end)
    ].copy()

    metrics = evaluate_baseline(
        train=train,
        test=test
    )

    baseline_results.append({
        "cutoff": cutoff,
        "train_rows": len(train),
        "test_rows": len(test),
        **metrics
    })

baseline_df = pl.DataFrame(baseline_results)

baseline_df
```

```python
def evaluate_moving_average(
    train: pd.DataFrame,
    test: pd.DataFrame,
    window: int = 7
) -> dict:

    history = train.copy()

    predictions = []

    for _, row in test.iterrows():

        last_values = history["y"].tail(window)

        prediction = last_values.mean()

        predictions.append(prediction)

        # Depois que o dia real acontece,
        # ele entra no histórico para a próxima previsão.
        history = pd.concat(
            [
                history,
                pd.DataFrame([row])
            ],
            ignore_index=True
        )

    y_true = test["y"].to_numpy()
    y_pred = np.array(predictions)

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mape = mean_absolute_percentage_error(
        y_true,
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "MAPE": mape
    }
```

```python
moving_average_results = []

for cutoff in cutoffs:

    train = df_prophet[
        df_prophet["ds"] <= cutoff
    ].copy()

    test_end = cutoff + pd.Timedelta(days=horizon)

    test = df_prophet[
        (df_prophet["ds"] > cutoff) &
        (df_prophet["ds"] <= test_end)
    ].copy()

    metrics = evaluate_moving_average(
        train=train,
        test=test,
        window=7
    )

    moving_average_results.append({
        "cutoff": cutoff,
        "train_rows": len(train),
        "test_rows": len(test),
        **metrics
    })

moving_average_df = pl.DataFrame(
    moving_average_results
)

moving_average_df
```

```python
def evaluate_moving_average_multistep(
    train: pd.DataFrame,
    test: pd.DataFrame,
    window: int = 7
) -> dict:

    # Histórico disponível antes do início do teste
    history = train["y"].tolist()

    predictions = []

    for _ in range(len(test)):

        # Últimos 7 valores disponíveis
        last_values = history[-window:]

        # Previsão
        prediction = np.mean(last_values)

        predictions.append(prediction)

        # IMPORTANTE:
        # adicionamos a PREVISÃO ao histórico,
        # e não o valor real
        history.append(prediction)

    y_true = test["y"].to_numpy()
    y_pred = np.array(predictions)

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mape = mean_absolute_percentage_error(
        y_true,
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "MAPE": mape
    }
    
```

```python
moving_average_results = []

for cutoff in cutoffs:

    train = df_prophet[
        df_prophet["ds"] <= cutoff
    ].copy()

    test_end = cutoff + pd.Timedelta(days=horizon)

    test = df_prophet[
        (df_prophet["ds"] > cutoff) &
        (df_prophet["ds"] <= test_end)
    ].copy()

    metrics = evaluate_moving_average_multistep(
        train=train,
        test=test,
        window=7
    )

    moving_average_results.append({
        "cutoff": cutoff,
        "train_rows": len(train),
        "test_rows": len(test),
        **metrics
    })

moving_average_multistep_df = pl.DataFrame(
    moving_average_results
)

moving_average_multistep_df
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing


def evaluate_holt_winters(
    train: pd.DataFrame,
    test: pd.DataFrame
) -> dict:

    model = ExponentialSmoothing(
        train["y"],
        trend=None,
        seasonal="add",
        seasonal_periods=7,
        initialization_method="estimated"
    )

    fitted_model = model.fit(
        optimized=True
    )

    forecast = fitted_model.forecast(
        len(test)
    )

    y_true = test["y"].to_numpy()
    y_pred = np.asarray(forecast)

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mape = mean_absolute_percentage_error(
        y_true,
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "MAPE": mape
    }
```

#### encapsular o backtesting do Holt-Winters

```python
def evaluate_holt_winters(df, cutoffs, horizon=28):

    results = []

    for cutoff in cutoffs:

        train = df[df["ds"] <= cutoff].copy()

        test = df[
            (df["ds"] > cutoff) &
            (df["ds"] <= cutoff + pd.Timedelta(days=horizon))
        ].copy()

        model = train_holt_winters(train)

        forecast = model.forecast(len(test)).to_numpy()

        y_true = test["y"].to_numpy()

        mae = mean_absolute_error(y_true, forecast)

        rmse = np.sqrt(
            mean_squared_error(y_true, forecast)
        )

        mape = mean_absolute_percentage_error(
            y_true,
            forecast
        )

        results.append({
            "cutoff": cutoff,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })

    return pd.DataFrame(results)
```

```python
def run_pipeline(
    df,
    cutoffs,
    forecast_end,
    dataset_name,
    model_dir="models"
):
    model_dir = Path(model_dir)
    model_dir.mkdir(exist_ok=True)

    # 1. Treinamento
    model = train_holt_winters(df)

    # 2. Backtesting
    metrics_df = evaluate_holt_winters(
        df=df,
        cutoffs=cutoffs
    )

    metrics = {
        "mae": float(metrics_df["MAE"].mean()),
        "rmse": float(metrics_df["RMSE"].mean()),
        "mape": float(metrics_df["MAPE"].mean())
    }

    # 3. Forecast
    forecast = generate_forecast(
        model=model,
        df=df,
        forecast_end=forecast_end
    )

    # 4. Metadata
    metadata = {
        "dataset": dataset_name,
        "model": "Holt-Winters",
        "trend": None,
        "seasonal": "add",
        "seasonal_periods": 7,

        "training_start": df["ds"].min().strftime("%Y-%m-%d"),
        "training_end": df["ds"].max().strftime("%Y-%m-%d"),

        "forecast_start": forecast["ds"].min().strftime("%Y-%m-%d"),
        "forecast_end": forecast["ds"].max().strftime("%Y-%m-%d"),

        "metrics": metrics
    }

    # 5. Salvar modelo
    model_path = model_dir / f"{dataset_name}_holt_winters.pkl"

    joblib.dump(
        model,
        model_path
    )

    # 6. Salvar metadata
    metadata_path = model_dir / f"{dataset_name}_metadata.json"

    with open(metadata_path, "w") as f:
        json.dump(
            metadata,
            f,
            indent=4
        )

    return {
        "model": model,
        "forecast": forecast,
        "metrics": metrics,
        "metadata": metadata
    }
```

# Prophet

```python
from prophet import Prophet


def train_prophet(df):
    model = Prophet(
        weekly_seasonality=True,
        daily_seasonality=False,
        yearly_seasonality=False
    )

    model.fit(
        df[["ds", "y"]]
    )

    return model
```

```python
def generate_prophet_forecast(model, df, forecast_end):

    training_end = df["ds"].max()

    future_dates = pd.date_range(
        start=training_end + pd.Timedelta(days=1),
        end=forecast_end,
        freq="D"
    )

    future = pd.DataFrame({
        "ds": future_dates
    })

    forecast = model.predict(future)

    return forecast[
        ["ds", "yhat", "yhat_lower", "yhat_upper"]
    ]
```

```python
def evaluate_prophet(df, cutoffs, horizon=28):

    results = []

    for cutoff in cutoffs:

        train = df[df["ds"] <= cutoff].copy()

        test = df[
            (df["ds"] > cutoff) &
            (df["ds"] <= cutoff + pd.Timedelta(days=horizon))
        ].copy()

        model = train_prophet(train)

        future = test[["ds"]].copy()

        forecast = model.predict(future)

        y_true = test["y"].to_numpy()
        y_pred = forecast["yhat"].to_numpy()

        mae = mean_absolute_error(y_true, y_pred)

        rmse = np.sqrt(
            mean_squared_error(y_true, y_pred)
        )

        mape = mean_absolute_percentage_error(
            y_true, y_pred
        )

        results.append({
            "cutoff": cutoff,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })

    return pd.DataFrame(results)
```

# Two Together

```python
def train_models(df):
    prophet_model = train_prophet(df)
    holt_winters_model = train_holt_winters(df)

    return {
        "prophet": prophet_model,
        "holt_winters": holt_winters_model
    }
```

```python
models = train_models(train_final)

models.keys()
```

```python
def generate_model_forecasts(models, df, forecast_end):

    forecasts = {}

    forecasts["prophet"] = generate_prophet_forecast(
        model=models["prophet"],
        df=df,
        forecast_end=forecast_end
    )

    forecasts["holt_winters"] = generate_forecast(
        model=models["holt_winters"],
        df=df,
        forecast_end=forecast_end
    )

    return forecasts
```

```python
forecasts = generate_model_forecasts(
    models=models,
    df=train_final,
    forecast_end="2026-12-31"
)
```

```python
def run_pipeline(
    df,
    cutoffs,
    forecast_end,
    dataset_name,
    model_dir="models"
):
    model_dir = Path(model_dir)
    model_dir.mkdir(exist_ok=True)

    # 1. Treinar
    models = train_models(df)

    # 2. Avaliar
    prophet_metrics = evaluate_prophet(
        df=df,
        cutoffs=cutoffs
    )

    hw_metrics = evaluate_holt_winters(
        df=df,
        cutoffs=cutoffs
    )

    # 3. Forecast
    forecasts = generate_model_forecasts(
        models=models,
        df=df,
        forecast_end=forecast_end
    )

    # 4. Métricas médias
    metrics = {
        "prophet": {
            "mae": float(prophet_metrics["MAE"].mean()),
            "rmse": float(prophet_metrics["RMSE"].mean()),
            "mape": float(prophet_metrics["MAPE"].mean())
        },
        "holt_winters": {
            "mae": float(hw_metrics["MAE"].mean()),
            "rmse": float(hw_metrics["RMSE"].mean()),
            "mape": float(hw_metrics["MAPE"].mean())
        }
    }

    # 5. Metadata
    metadata = {
        "dataset": dataset_name,

        "training_start": df["ds"].min().strftime("%Y-%m-%d"),
        "training_end": df["ds"].max().strftime("%Y-%m-%d"),

        "forecast_start": forecasts["prophet"]["ds"].min().strftime("%Y-%m-%d"),
        "forecast_end": forecasts["prophet"]["ds"].max().strftime("%Y-%m-%d"),

        "models": {
            "prophet": {
                "weekly_seasonality": True,
                "daily_seasonality": False,
                "yearly_seasonality": False
            },
            "holt_winters": {
                "trend": None,
                "seasonal": "add",
                "seasonal_periods": 7
            }
        },

        "metrics": metrics
    }

    # 6. Salvar modelos
    joblib.dump(
        models["prophet"],
        model_dir / f"{dataset_name}_prophet.pkl"
    )

    joblib.dump(
        models["holt_winters"],
        model_dir / f"{dataset_name}_holt_winters.pkl"
    )

    # 7. Salvar metadata
    with open(
        model_dir / f"{dataset_name}_metadata.json",
        "w"
    ) as f:
        json.dump(
            metadata,
            f,
            indent=4
        )

    return {
        "models": models,
        "forecasts": forecasts,
        "metrics": metrics,
        "metadata": metadata
    }
```

```python
result = run_pipeline(
    df=train_final,
    cutoffs=cutoffs,
    forecast_end="2026-12-31",
    dataset_name="sales"
)
```
